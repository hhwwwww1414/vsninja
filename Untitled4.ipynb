```python
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/dianasusla28/andan/refs/heads/main/data_sales.csv', parse_dates=['order_date'])
p = pd.read_csv('https://raw.githubusercontent.com/dianasusla28/andan/refs/heads/main/payment_info.csv')
```

```python
(df.order_date.dt.month == 8).sum()
```

```python
n = df[df.region == 'North']
n.loc[n.revenue.idxmin(), 'discount']
```

```python
q1, q2, q3 = df.price.quantile([.25, .5, .75])
df['coded'] = df.price.apply(lambda x: 111 if x <= q1 else 222 if x <= q2 else 333 if x <= q3 else 444)
round(df[df.order_date.dt.hour >= 12].coded.var(), 2)
```

```python
m = df.merge(p, on='order_id')
top = m.payment_method.mode()[0]
m[m.payment_method == top].quantity.sum()
```

```python
m['c'] = m.shipping_cost / m.quantity
m.loc[m.c.idxmax(), 'order_date'].date()
```

```python
m.groupby(m.order_date.dt.date)[['quantity', 'shipping_cost']].sum().plot.scatter('quantity', 'shipping_cost')
```
